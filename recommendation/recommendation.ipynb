```python
import requests
from transformers import pipeline
from langchain.prompts import PromptTemplate
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.schema.runnable import RunnablePassthrough
import pandas as pd
import chromadb
from langchain.embeddings import OpenAIEmbeddings
from langchain.chat_models import ChatOpenAI
from chromadb.utils import embedding_functions
import os
api_key = "api key"
os.environ["OPENAI_API_KEY"] = api_key
```

```python
# 프롬프트 템플릿 설정
template = """당신은 주어진 검색 결과와 사용자의 질문을 바탕으로 답변하는 AI 어시스턴트입니다. 당신은 취얼업 팀에 의해 만들어졌습니다.
존댓말로 정중하게 답변하세요.

사용자의 희망 직무, 경력, 학위, 사용 기술, 프로젝트 경험을 고려하여 적합한 공고를 최대 5개 추천해 주세요.
각 공고의 **URL, 직무명, 요구 기술 스택, 학위 요구 사항, 요구 언어, 프레임워크, 라이브러리, 툴**을 포함해야 합니다.
공고 내용의 경우 요약해서 출력하세요.

**출력 형식 (예시)**
1. 직무명:
    - 공고 내용:
    - URL:
    - 경력 요구 사항:
    - 학위 요구 사항:
    - 요구 언어:
    - 요구 프레임워크:
    - 요구 라이브러리:
    - 요구 툴:

2. 직무명:
    - 공고 내용:
    - URL:
    - 경력 요구 사항:
    - 학위 요구 사항:
    - 요구 언어:
    - 요구 프레임워크:
    - 요구 라이브러리:
    - 요구 툴:

검색 결과가 없으면 **"현재 적절한 공고가 없습니다."**라고 답변하세요.

{context}

**사용자 질문**
{input}

**추천 공고:**
"""

# PromptTemplate 초기화
prompt = PromptTemplate(
    template=template,
    input_variables=['context', 'input']  # 'input'을 필수로 추가
)

# ✅ LLM (GPT-3.5) 모델 설정
llm = ChatOpenAI(model_name='gpt-3.5-turbo', temperature=0)

# ✅ LLM 체인 설정
llm_chain = prompt | llm
```

```python
# 데이터 로드
df = pd.read_csv('./2025-01-24-12_categorized.csv', encoding='utf-8-sig')

# 필요한 컬럼 선택
df = df[['id', 'title', 'url', 'duty', 'combined_text', 'career', 'degree', 'language', 'it_language', 'framework', 'library', 'tool']]
df.index = range(0, len(df))
df[['it_language', 'framework', 'library', 'tool']] = df[['it_language', 'framework', 'library', 'tool']].fillna('').astype(str)

def get_recommendations(df, duty, career, degree, project,
                     it_language=None, framework=None, library=None, tool=None, top_k: int = 5):

    # 로컬 저장 경로 설정 (예: Google Drive에 저장)
    DB_PATH = "./chroma_db"

    # 저장된 ChromaDB 불러오기
    chroma_client = chromadb.PersistentClient(path=DB_PATH)

    # 임베딩 함수 객체 생성
    emb_model = embedding_functions.OpenAIEmbeddingFunction(api_key=api_key)

    collection = chroma_client.get_or_create_collection(
        name="job_postings",
        embedding_function=emb_model,
        metadata={
            "hnsw:space": "cosine",
            "hnsw:search_ef": 200
        }
    )

    filtered_df = df.loc[:].copy()

    def filter_list_column(df, column, values):
        if values:
            return df[df[column].str.split(", ").apply(lambda x: any(lang in values for lang in x))]
        return df

    if duty:
        filtered_df = filtered_df[filtered_df['duty'].str.contains(duty, case=False, na=False)]  # duty 문자열 포함 여부 체크
    # filtered_df = filter_list_column(filtered_df, 'duty', duty)

    if career is not None:
        filtered_df = filtered_df[filtered_df['career'] == career]

    if degree:
        filtered_df = filtered_df[filtered_df['degree'] == degree]


    filtered_df = filter_list_column(filtered_df, 'it_language', it_language)
    filtered_df = filter_list_column(filtered_df, 'framework', framework)
    filtered_df = filter_list_column(filtered_df, 'library', library)
    filtered_df = filter_list_column(filtered_df, 'tool', tool)

    id_list = filtered_df.index.to_list()

    if id_list == []:
        x = "현재 적절한 공고가 없습니다."
        print(x)
        return None

    else:
        # OR 조건 생성
        where_clause = {"$or": [{"id": id} for id in id_list]}

        # 검색 쿼리 생성 (프로젝트 내용을 포함하여 검색)
        query = f"프로젝트 경험: {project}"

        # ChromaDB에서 검색 실행
        results = collection.query(
            query_texts= query,  # 검색어 벡터화
            n_results=top_k,
            where = where_clause  # 필터 적용
        )

        # 검색 결과를 context로 변환
        context = ""
        for i in range(len(results['metadatas'][0])):
            res = results['metadatas'][0][i]
            doc = results['documents'][0][i]
            formatted_output = f"""
            직무명: {res['duty']}
            - 공고 내용: {doc}\n
            - URL: {res['url']}
            - 경력 요구 사항: {res['career']}년 이상
            - 학위 요구 사항: {res['degree']}
            - 요구 언어: {res.get('it_language', 'N/A')}
            - 요구 프레임워크: {res.get('framework', 'N/A')}
            - 요구 라이브러리: {res.get('library', 'N/A')}
            - 요구 툴: {res.get('tool', 'N/A')}\n
            """
            context += formatted_output

    # LLM 체인 실행
    return llm_chain.invoke({"context": context, "input": project})
```

```python
# 사용 예시
result = get_recommendations(df, duty = "AI", career = 0, degree = 1, project = "CV 관련 프로젝트 경험", it_language = ["python"])

if result == "":
    print(result)
else:
    print(result.content)
```
