```
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import ast
import numpy as np
```

자격요건

```
# CSV 파일 읽기
df = pd.read_csv('/content/detailed_tech_stack_check_with_columns (1).csv')  # 파일 이름 변경 필요

# 특정 열 가져오기 (예: 'column_name' 열의 텍스트만 가져오기)
combine_text_list = df['combined_analyzed'].tolist()
combine_string_list = []


# 각 항목의 문자열을 평가(eval)하고, 띄어쓰기로 구분된 문자열로 변환
for combine in combine_text_list:
    # 문자열을 Python 리스트로 변환 (eval 사용)
    combine = eval(combine)
    # 띄어쓰기로 구분된 문자열로 변환
    combine = " ".join(combine)
    combine_string_list.append(f"[{combine}]")



# TF-IDF 계산
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(combine_string_list)

# 단어 목록 및 TF-IDF 행렬 출력
feature_names = vectorizer.get_feature_names_out()  # 단어 목록
tfidf_array = X.toarray()  # TF-IDF 결과를 numpy array로 변환


# DataFrame으로 변환(TF-IDF 원본)
tfidf_df = pd.DataFrame(tfidf_array, columns=feature_names)
```

```
# 열별 합계 계산
summed_tfidf = np.sum(tfidf_array, axis=0)

# DataFrame 생성 (하나의 row로 합친 것)
summed_tfidf_df = pd.DataFrame([summed_tfidf], columns=feature_names)

# Row-column 전환 및 내림차순 정렬
summed_tfidf_df = summed_tfidf_df.T.sort_values(by=0, ascending=False)

# 8. 결과 출력
print("내림차순으로 정렬한 TF-IDF 값:\n", summed_tfidf_df)
```

```
# 집계된 결과를 CSV 파일로 저장
output_csv_path = "/tfidf_combine_sum_matrix.csv"
summed_tfidf_df.to_csv(output_csv_path, index=True, encoding="utf-8-sig")

print(f"집계된 TF-IDF 결과가 CSV 파일로 저장되었습니다: {output_csv_path}")
```
