```python
### 필요한 라이브러리 / 함수 임폴트
import os
from google.colab import userdata
import pandas as pd
import ast
```

```python
# CSV 파일 읽기
df = pd.read_csv('/content/updated_4way_classification.csv')  # 파일 이름 변경 필요

# 특정 열 가져오기 (예: 'column_name' 열의 텍스트만 가져오기)
id_list = df['id'].tolist()
oov_text_list = df['combine_oov'].tolist()
oov_text_list2 = df['pre_oov'].tolist()


print(oov_text_list)
```

```python
converted_2d_list = []
converted_2d_list2 = []
# 문자열을 실제 2차원 리스트로 변환
converted_2d_list = [ast.literal_eval(item) for item in oov_text_list]
converted_2d_list2 = [ast.literal_eval(item) for item in oov_text_list2]


oov_list = []
for oov in converted_2d_list:
  for item in oov:
    oov_list.append(item)
for oov in converted_2d_list2:
  for item in oov:
    oov_list.append(item)

# 중복 제거 및 정렬
unique_oov_list = set(oov_list)

# 음절이 하나인 토큰 제거
filtered_tokens = {token for token in unique_oov_list if len(token) > 1}

# 결과 출력
print("중복제거전:",len(oov_list))
# 결과 출력
print("중복제거후:",len(unique_oov_list))
# 결과 출력
print("음절하나 제거후:",len(filtered_tokens))
```
