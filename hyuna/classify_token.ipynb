```python
# 이중 리스트를 CSV 파일로 저장 (각각 컬럼으로 넣기)
def save_to_csv_columns(file_name, columns, column_names):
    with open(file_name, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow(column_names)  # 헤더 추가
        max_len = max(len(col) for col in columns)
        for i in range(max_len):
            row = [columns[0][i] if i < len(columns[0]) else ""] + [", ".join(columns[col][i]) if i < len(columns[col]) else "" for col in range(1, len(columns))]
            writer.writerow(row)
# CSV 파일로 저장
save_to_csv_columns("data_columns.csv", [combine_id_list, combine_language_list, pre_language_list,combine_framework_list, pre_framework_list, combine_libraries_list, pre_libraries_list, combine_tools_list, pre_tools_list, combine_oov_list, pre_oov_list], ["id","combine_languages", "pre_languages","combine_frameworks", "pre_frameworks", "combine_libraries","pre_libraries", "combine_tools", "pre_tools", "combine_oov", "pre_oov"])

# 출력 결과 확인
print("Data saved to data_columns.csv")
```

```python
#가윤님 동의어 사전과 합쳐서 json으로 저장
import json

# Load original JSON file
with open('final_synonym_corrected.json', 'r', encoding='utf-8') as file:
    original_json = json.load(file)

# Define synonym data
synonym_data = {
    'language': {},
    'frameworks': {
        'backbonejs': 'backbone js',
        'emberjs': 'ember js',
        'expressjs': 'express js',
        'nestjs': 'nest js',
        'nextjs': 'next js',
        'nuxtjs': 'nuxt js',
        'vuejs': 'vue js',
        'nodejs': 'node js',
    },
    'libraries': {
        'reactjs': 'react js',
    },
    'tools': {
        'arangodb': 'arango db',
        'aws auroradb': 'aws aurora db',
        'aws documentdb': 'aws document db',
        'aws dynamodb': 'aws dynamo db',
        'aws mariadb': 'aws maria db',
        'cassandradb': 'cassandra db',
        'cockroachdb': 'cockroach db',
        'mongodb': 'mongo db',
        'mssql': 'ms sql',
        'mysql': 'my sql',
        'oracledb': 'oracle db',
        'postgresql': 'postgre sql',
        'rocksdb': 'rocks db',
    }
}

# Merge synonym data into original JSON
for category, items in synonym_data.items():
    if category in original_json:
        for key, value in items.items():
            if key in original_json[category]:
                if isinstance(original_json[category][key], list):
                    original_json[category][key].append(value)
                else:
                    original_json[category][key] = [original_json[category][key], value]

# Save the merged JSON
with open('merged_synonym_corrected.json', 'w', encoding='utf-8') as file:
    json.dump(original_json, file, ensure_ascii=False, indent=4)

print("Merged JSON saved as 'merged_synonym_corrected.json'")
```
