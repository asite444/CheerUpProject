```python
### 필요한 라이브러리 / 함수 임폴트
import os
from langchain_openai import ChatOpenAI
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.document_loaders.csv_loader import CSVLoader
from langchain.prompts import PromptTemplate
from google.colab import userdata
from langchain.chains import RetrievalQA
import pandas as pd
import ast
import numpy as np
import csv
```

```python
## openai api key 불러오기
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

# 객체 생성
llm = ChatOpenAI(
    temperature=0.1,  # 창의성 (0.0 ~ 2.0)
    max_tokens=2048,  # 최대 토큰수
    model_name="gpt-3.5-turbo",  # 모델명
)
```

```python

# loader = CSVLoader(file_path='/content/2025-01-24-10_categorized.csv', encoding='cp949')
# data = loader.load()
df = pd.read_csv('/content/2025-01-24-10_categorized.csv')  # 파일 이름 변경 필요

# 특정 열 가져오기 (예: 'column_name' 열의 텍스트만 가져오기)
combine_id_list = df['json'].tolist()
json_top_ten = []

for i in range(0,10):
  json_top_ten.append(combine_id_list[i])
  print(combine_id_list[i])

# 새로운 데이터베이스 생성
test_db = Chroma.from_documents(
    json_top_ten,
    HuggingFaceEmbeddings(
        model_name="sentence-transformers/xlm-r-100langs-bert-base-nli-stsb-mean-tokens",
        model_kwargs={'device': 'cuda'},
        encode_kwargs={'normalize_embeddings': True}
    )
)
```

```python
print(test_db._collection.count())

question = "리액트와 리덕트를 할 수 있는 경우 지원 가능한 공고가 뭐가 있을까?"
top_five_post = test_db.similarity_search_with_relevance_scores(question,k=5)

for doc in top_five_post:
  print(doc)
  print('--'*10)
```

```python
retriever = test_db.as_retriever(search_kwargs={"k": 10})

# Create Prompt
template = """당신은 개발 직군의 채용공고를 보고 현재 트랜드를 분석해주는 리포터입니다.
주어진 검색 결과를 바탕으로 답변하세요.
{context}

Question: {question}
Answer:
"""

prompt = PromptTemplate.from_template(template)

qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type_kwargs={"prompt": prompt},
    retriever=retriever,
    return_source_documents=True)
```

```python
# input_text = "개발 분야에서 채용 공고가 가장많은 직군을 9개까지 알려줘"
# input_text = "프론트엔드 개발자에서 백엔드 개발자가 되기 위해 어떤 프로젝트 경험이 있어야 하는지 채용공고를 참고해서 알려줘"
# input_text = "나는 백엔드 개발자로 경력이 4년 있고, 노드js랑 네스트js, 그리고 RESTFULT API도 많이 사용해봤어. 관련도가 높은 백엔드 채용공고 5개의 url을 이유와 함께 출력해줘. 방금 말한 조건에서 지역이 서울인 것만"
input_text = "프론트엔드 개발자랑 업무 전환이 가장 용이한 직무를 이유와 함께 알려줘"
chatbot_response = qa_chain.invoke(input_text)
print(chatbot_response["result"])
```
