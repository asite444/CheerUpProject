```python
import mecab
import pandas as pd
```

```python
mc = mecab.MeCab()


# CSV 파일 읽기
df = pd.read_csv('/content/merged_df.csv')  # 파일 이름 변경 필요

# 특정 열 가져오기 (예: 'column_name' 열의 텍스트만 가져오기)
re_text_list = df['description']+ ' '+ df['requirement']+ ' '+ df['technicalTags']
re_text_list = re_text_list.tolist()
pre_text_list = df['preferredExperience'].tolist()

# technicalTags의 토큰
# 요구사항과 자격요건의 토큰
requirement_tokens = []
for requirement in re_text_list:
  if pd.isna(requirement) or not isinstance(requirement, str):
        tokens = []
  else :
    re_pos = mc.pos(requirement) # 형태소의 리턴인 리스트에 품사 태깅 return은 튜플
    tokens = [word for word, pos in re_pos if pos in ['SL', 'NNG', 'NNP']]#명사랑 영어만 추출
    # print(tokens)
  requirement_tokens.append(tokens)
df['requirement_tokens'] = [','.join(row) for row in requirement_tokens]
df.to_csv('/content/merged_df.csv', index=False)


# 우대사항의 토큰
prefferedEx_tokens = []
for preferredEx in pre_text_list:
  if pd.isna(preferredEx) or not isinstance(preferredEx, str):
        tokens = []
  else:
    pos = mc.pos(preferredEx) # 형태소의 리턴인 리스트에 품사 태깅 return은 튜플
    tokens = [word for word, pos in pos if pos in ['SL', 'NNG', 'NNP']]#명사랑 영어만 추출
  prefferedEx_tokens.append(tokens)
df['prefferedEx_tokens'] = [','.join(row) for row in prefferedEx_tokens]
df.to_csv('/content/merged_df.csv', index=False)

print(requirement_tokens)
print(prefferedEx_tokens)
```
